```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load dataset
df = pd.read_csv("../data/teachers.csv")

df.head()
```

```python
def assign_performance(row):
    if (row['attendance_percentage'] >= 95 and
        row['student_rating'] >= 4.5 and
        row['sentiment_score'] > 0.3):
        return "Excellent"

    elif (row['attendance_percentage'] >= 90 and
          row['student_rating'] >= 4.0):
        return "Good"

    else:
        return "Average"


df['performance'] = df.apply(assign_performance, axis=1)

df[['attendance_percentage','student_rating','performance']].head()
```

```python
print(df['performance'].value_counts())
```

```python
X = df[
    ['attendance_percentage',
     'experience_years',
     'student_rating',
     'workload_hours',
     'sentiment_score']
]

y = df['performance']
```

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

y_encoded[:5]
```

```python
print(label_encoder.classes_)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.3, random_state=42
)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))
```

```python
# New teacher data
new_teacher = pd.DataFrame({
    'attendance_percentage': [88],
    'experience_years': [4],
    'student_rating': [4.3],
    'workload_hours': [40],
    'sentiment_score': [0.2]
})

prediction = model.predict(new_teacher)
predicted_label = label_encoder.inverse_transform(prediction)

print("Predicted Performance:", predicted_label[0])
```

```python
import joblib

joblib.dump(model, "performance_model.pkl")
print("Model saved successfully")
```

```python
loaded_model = joblib.load("performance_model.pkl")

prediction = loaded_model.predict(new_teacher)
predicted_label = label_encoder.inverse_transform(prediction)

print("Predicted Performance (Loaded Model):", predicted_label[0])
```
